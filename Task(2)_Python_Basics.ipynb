(3)Write a Python program that takes two numbers as input and prints their sum and difference.

Example:
Input:  10, 5  
Output:  
Sum: 15  
Difference: 5  


```python
num1=int(input("enter num1"))
num2=int(input("enter num2 "))
sum= num1 + num2
print('sum : ',sum)
diff= num1-num2
print('diff : ',diff)
```
